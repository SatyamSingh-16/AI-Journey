```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [21, 25, np.nan, 35, 40],
    "Salary": [30000, 45000, 60000, np.nan, 100000],
    "Department": ["IT", "HR", "IT", "Finance", "IT"]
})

print(df["Department"].unique())
df["Department"].nunique()
print(df.isnull())
print(df.isnull().sum())
df.isna()
df.dropna()
df.duplicated()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [30000, 45000, 60000, 75000, 50000, 80000],
    "Experience": [1, 2, 5, 8, 3, 7]
})

print(df)

print(df[df["Department"]=="IT"]["Salary"].mean())
print(df[df["Department"]=="HR"]["Salary"].mean())
print(df[df["Department"]=="Finance"]["Salary"].mean())

df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].agg(["mean","min","max","std"])
```

```python
import numpy as np
import pandas as pd

customers = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5],
    "Name": ["A", "B", "C", "D", "E"],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Bangalore"]
})

orders = pd.DataFrame({
    "CustomerID": [1, 2, 2, 3, 6],
    "Amount": [500, 1000, 700, 1200, 900]
})

print(pd.merge(customers,orders,on="CustomerID",how="inner"))
print("----------------------------------------")
print(pd.merge(customers,orders,on="CustomerID",how="left"))
print("----------------------------------------")
print(pd.merge(customers,orders,on="CustomerID",how="outer"))
print("----------------------------------------")
left = pd.merge(customers,orders,on="CustomerID",how="left")
print(left[left["Amount"].isnull()])
print("----------------------------------------")
inner = pd.merge(customers,orders,on="CustomerID",how="inner")
print(inner.groupby("CustomerID")["Amount"].sum())
```

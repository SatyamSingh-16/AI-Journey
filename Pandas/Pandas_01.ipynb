```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [21, 25, 30, 35, 40],
    "Salary": [30000, 45000, 60000, 75000, 100000],
    "Department": ["IT", "HR", "IT", "Finance", "IT"]
})

print(df.head(5))
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print("---------------------------------------------")
print(df["Salary"])
print(df[["Salary","Name"]])
print(df.iloc[2])
print(df.loc[2,"Salary"])
print("---------------------------------------------")
print(df[df["Salary"] > 50000])
print(df[df["Department"] == "IT"])
print(df[(df["Salary"] > 50000) & (df["Department"] == "IT")])
```

```python
import matplotlib.pyplot as plt

experience = [1,2,3,4,5]
salary = [30000, 35000, 42000, 50000,60000]

fig,ax = plt.subplots()
ax.plot(experience, salary)
ax.set_title("Experience vs Salary")
ax.set_xlabel("Years of Experience")
ax.set_ylabel("Salary")

plt.show()
```

```python
import matplotlib.pyplot as plt

experience = [1,2,3,4,5]
salary = [30000, 35000, 42000, 50000,60000]

fig,ax = plt.subplots()
ax.scatter(experience, salary)
ax.set_title("Experience vs Salary")
ax.set_xlabel("Years of Experience")
ax.set_ylabel("Salary")

plt.show()
```

```python
import matplotlib.pyplot as plt

salary = [25000, 30000, 32000, 35000, 40000,
          45000, 50000, 60000, 80000, 150000]

fig, ax = plt.subplots()

ax.hist(salary)

ax.set_title("Salary Distribution")
ax.set_xlabel("Salary")
ax.set_ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt

salary = [25000, 30000, 32000, 35000, 40000,
          45000, 50000, 60000, 80000, 150000]

fig, ax = plt.subplots()

ax.boxplot(salary)

ax.set_title("Salary Distribution")
ax.set_xlabel("Salary")
ax.set_ylabel("Frequency")

plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [21, 25, 28, 32, 35, 42, 26, 30],
    "Salary": [25000, 40000, 55000, 70000, 90000, 120000, 50000, 65000],
    "Experience": [0, 2, 4, 7, 10, 18, 3, 6],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "IT", "Finance", "HR", "IT"
    ]
})

fig,ax = plt.subplots()
ax.hist(df["Salary"])

ax.set_title("Salary Distribution")
ax.set_xlabel("Salary")
ax.set_ylabel("Frequency")

plt.show()

print("----------------------------------------------")

ax.scatter(df["Experience"],df["Salary"])

plt.show()
```

```python
import numpy as np

X = np.array([
    [20, 30000],
    [25, 40000],
    [30, 60000],
    [35, 80000],
    [40, 100000]
])

print(X.shape)
print(X.ndim)
print(X.size)
age = X[:,0]
print(np.mean(age))
salary = X[:,1]
print(np.mean(salary))

print(np.max(salary))
print(np.argmax(salary))

res = np.where(salary > 50000,"G","S")
print(res)

res1 = age - np.mean(age)
res2 = res1 / np.std(age)
print(res2)
```

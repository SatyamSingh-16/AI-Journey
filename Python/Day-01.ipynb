```python
import numpy as np
students = np.array([
    [85, 78, 92],
    [67, 88, 76],
    [90, 95, 89],
    [72, 65, 80]
])

print(students.ndim)
print(students.shape)
print(students.size)
print(students.dtype)

print(np.mean(students,axis=0))
print(np.mean(students,axis=1))
print(np.argmax(students))
print(students[students > 85])

```
